```python
data = {
    'Name': ['Rahul', 'Priya', 'Aman', 'Sneha', 'Rohan', 'Anjali'],
    'Age': [23, None, 28, 31, None, 26],
    'Salary': [40000, 55000, None, 70000, 65000, None],
    'Experience': [1, 3, 5, None, 4, 2]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df.isnull().sum()
```

```python
df["Age"].fillna(df["Age"].mean(),inplace=True)
df["Salary"].fillna(df["Salary"].median(),inplace=True)
df["Experience"].fillna(df["Experience"].mode(),inplace=True)
```

```python
print(df)
```

**ENCODING LABELLED DATA**

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

df = pd.read_csv("sample_data.csv")

df_label = df.copy()

le = LabelEncoder()
df_label["Gender_Encoded"] = le.fit_transform(df_label["Gender"])
df_label["Passed_Encoded"] = le.fit_transform(df_label["Passed"])
```

```python
print("Label Encoded Data:")
print(df_label[["Name","Gender","Gender_Encoded","Passed","Passed_Encoded"]])
```
