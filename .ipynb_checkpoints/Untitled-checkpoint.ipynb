```python
import pandas as pd
```

```python
data = {
    'Name': ['Pavan', 'Kapil', 'Lalit', 'Ishan', 'Om'],
    'Age': [25, None, 44, 23, None],
    'Salary': [50000, 60000, 70000, None, None]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Original DataFrame")
```

```python
print(data)
```

```python
print(df.isnull().sum())
```

```python
df_drop = df.dropna()
```

```python
print(df_drop)
```

```python
df["Age"].fillna(df["Age"].mean(),inplace=True)
df["Salary"].fillna(df["Salary"].mean(),inplace=True)
print(df)
```

```python
df.isnull().mean()*100
```

```python
data = {
    'Name': ['Rahul', 'Priya', 'Aman', 'Sneha', 'Rohan', 'Anjali'],
    'Age': [23, None, 28, 31, None, 26],
    'Salary': [40000, 55000, None, 70000, 65000, None],
    'Experience': [1, 3, 5, None, 4, 2]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df.isnull().sum()
```

```python
df["Age"].fillna(df["Age"].mean(),inplace=True)
df["Salary"].fillna(df["Salary"].median(),inplace=True)
df["Experience"].fillna(df["Experience"].mode(),inplace=True)
```

```python
print(df)
```

```python
df.isnull().sum()
```
