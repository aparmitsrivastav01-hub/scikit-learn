```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()

X = [[1], [2], [3], [4], [5]]
y = [40, 50, 65, 75, 90]

```

```python
model.fit(X,y)
value = float(input("Enter Hours Studied"))
print(model.predict([[value]]))
```
